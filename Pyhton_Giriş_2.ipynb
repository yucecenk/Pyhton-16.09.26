```python

kilo = float(input("Kilonuzu giriniz (kg): "))
boy_cm = float(input("Boyunuzu cm cinsinden giriniz (örn: 175): "))


boy_m = boy_cm / 100


bmi = kilo / (boy_m ** 2)


print(f"\nVücut Kitle İndeksiniz (BMI): {bmi:.2f}")


if bmi < 18.5:
    print("Durum: Zayıf")
elif 18.5 <= bmi < 25:
    print("Durum: Normal (İdeal ağırlık)")
elif 25 <= bmi < 30:
    print("Durum: Kilolu (Fazla kilolu)")
else:
    print("Durum: Obez")
```

```python

notlar = []

print("Lütfen 4 dersin notunu sırasıyla giriniz (0-100 arası):")


for i in range(1, 5):
    not_degeri = float(input(f"{i}. dersin notu: "))
    notlar.append(not_degeri)


toplam = sum(notlar)
ortalama = toplam / len(notlar)


en_yuksek = notlar[0]
en_dusuk = notlar[0]

for not_degeri in notlar:
    if not_degeri > en_yuksek:
        en_yuksek = not_degeri
    if not_degeri < en_dusuk:
        en_dusuk = not_degeri


if 90 <= ortalama <= 100:
    harf_notu = "AA"
elif 85 <= ortalama < 90:
    harf_notu = "BA"
elif 75 <= ortalama < 85:
    harf_notu = "BB"
elif 70 <= ortalama < 75:
    harf_notu = "CB"
elif 60 <= ortalama < 70:
    harf_notu = "CC"
elif 50 <= ortalama < 60:
    harf_notu = "DC"
else:
    harf_notu = "FF (Kaldınız)"


print("\n--- ÖĞRENCİ NOT ANALİZİ ---")
print(f"Girilen Notlar: {notlar}")
print(f"Genel Ortalama: {ortalama:.2f}")
print(f"Harf Notunuz: {harf_notu}")
print(f"En Yüksek Not: {en_yuksek}")
print(f"En Düşük Not: {en_dusuk}")
```

```python

sepet = [45.50, 120.00, 80.00, 250.00, 65.00]

print(f"Sepetteki Tüm Ürünlerin Fiyatları: {sepet}")


pahali_urunler = []

for fiyat in sepet:
    if fiyat > 100:
        pahali_urunler.append(fiyat)


toplam_tutar = sum(sepet)


indirim_orani = 0.10
odenecek_tutar = toplam_tutar

if toplam_tutar > 500:
    indirim_miktari = toplam_tutar * indirim_orani
    odenecek_tutar = toplam_tutar - indirim_miktari
    indirim_mesaji = "Tebrikler! 500 TL üzeri alışveriş yaptığınız için %10 indirim kazandınız."
else:
    indirim_mesaji = "500 TL altı alışverişlerde indirim uygulanmamaktadır."


print("\n--- ALIŞVERİŞ SEPETİ ÖZETİ ---")
print(f"100 TL Üzeri Olan Ürünler: {pahali_urunler}")
print(f"Sepet Toplamı: {toplam_tutar:.2f} TL")
print(indirim_mesaji)

if toplam_tutar > 500:
    print(f"İndirim Miktarı: {indirim_miktari:.2f} TL")
    print(f"Ödemeniz Gereken Tutar: {odenecek_tutar:.2f} TL")
else:
    print(f"Ödemeniz Gereken Tutar: {odenecek_tutar:.2f} TL")
```
