```python
tam_sayi = 42
ondalikli_sayi = 3.14
yazi = "Python"
durum = True

# type() fonksiyonu ile her birinin tipini ekrana yazdırıyoruz
print(type(tam_sayi))        # Çıktı: <class 'int'>
print(type(ondalikli_sayi))  # Çıktı: <class 'float'>
print(type(yazi))            # Çıktı: <class 'str'>
print(type(durum))           # Çıktı: <class 'bool'>
print(type(10))    # Çıktı: <class 'int'>
print(type("10"))  # Çıktı: <class 'str'>
```
