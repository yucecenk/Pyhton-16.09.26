```python
class Araba:
    def __init__(self, marka, model, hiz=0):
        self.marka = marka
        self.model = model
        self.hiz = hiz

    def hizlandir(self, artis):
        self.hiz += artis
        print(f"{self.marka} {self.model} hızlandı. Yeni hız: {self.hiz} km/s")

    def bilgi_goster(self):
        print("\n--- Araç Bilgileri ---")
        print(f"Marka: {self.marka}")
        print(f"Model: {self.model}")
        print(f"Hız: {self.hiz} km/s")


arac1 = Araba("Mercedes", "EQB")


arac1.bilgi_goster()


arac1.hizlandir(50)
arac1.hizlandir(30)

arac1.bilgi_goster()
```

```python
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.uzunluk = uzunluk
        self.genislik = genislik

    def alan_hesapla(self):
        return self.uzunluk * self.genislik

    def cevre_hesapla(self):
        return 2 * (self.uzunluk + self.genislik)


dikdortgen1 = Dikdortgen(10, 5)

print(f"Dikdörtgenin Alanı: {dikdortgen1.alan_hesapla()}")
print(f"Dikdörtgenin Çevresi: {dikdortgen1.cevre_hesapla()}")
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa_sayisi):
        self.ad = ad
        self.yazar = yazar
        self.sayfa_sayisi = sayfa_sayisi

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"


kitap1 = Kitap("Suç ve Ceza", "Dostoyevski", 671)

print(kitap1.ozet())
```
